```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
import seaborn as sns


np.random.seed(42)

X, y = load_iris(return_X_y=True)

iris_df = pd.DataFrame(X, columns=["sepal length (cm)",
                                   "sepal width (cm)", 
                                   "petal length (cm)", 
                                   "petal width (cm)",])
iris_df["species"] = y
iris_df


```

```python
sns.heatmap(iris_df.drop("species", axis='columns').corr(), cmap="coolwarm", annot=True)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

iris_X_df = iris_df.drop("species", axis='columns')

features = iris_df.drop(columns="species").columns   # alla kolumner utom species


for i, feature in enumerate(features):
    sns.boxplot(
        data=iris_df,
        x="species",
        y=feature,
        hue="species",
        palette="deep",
        legend=True,
        ax=axes[i]
    )

plt.tight_layout()
plt.show()
```
